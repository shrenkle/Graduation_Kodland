```python
x = 0
y = 0
battery = 10

while True:
    print("Robby Path Finding Robot Simulation Program")
    print("[1] Location \n[2] Move \n[3] Reset \n[4] Quit \n[5] Battery Check \n[6] Recharge")
    try:
        choice = int(input("Enter choice >>"))
    except ValueError:
        print("Enter a valid choice number")
        continue

    if choice == 1:
        print(f'Robby is currently at ({x},{y})')
    elif choice == 2:
        direction = ""
        distance = 0
        while direction not in ["N","E","W","S"]:
            direction = input("Enter direction (N,E,W,S) >>").upper()
        while distance <= 0:
            try:
                distance = float(input("Enter distance >>"))
            except ValueError:
                print("Enter a valid number")
                continue


        if distance > battery:
            print(f"Not enough battery! That requires {distance}% battery, but you only have {battery}%")
        else:
            battery -= distance
            print(f'Robby moved {direction} | {distance} units. Battery left: {battery}%')
            if direction == 'N':
                y += distance
            elif direction == "E":
                x += distance
            elif direction == "W":
                x -= distance
            elif direction == "S":
                y -= distance
    elif choice == 3:
        x = y = 0
        print(f'Robby reset at ({x},{y})')
    elif choice == 4:
        break
    elif choice == 5:
        print(f"Current Battery: {battery}%")
    elif choice == 6:
        battery = 10
        print("Battery fully recharged to 10%")

print("Program Terminated")
```
